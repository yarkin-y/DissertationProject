```python
def nan_analysis(df):
    """
    This function takes a DataFrame and returns a DataFrame with the percentage of NaN values for each column.
    Only columns with NaN values are included in the output DataFrame.
    """
    nan_percentage = df.isna().mean() * 100
    nan_percentage = nan_percentage[nan_percentage > 0].reset_index()
    nan_percentage.columns = ['Column Name', 'Percentage of NaN Values']
    nan_percentage = nan_percentage.sort_values(by='Percentage of NaN Values', ascending=False)
    return nan_percentage
```

```python

def plot_correlation_matrix(df):
    """
    Plots a correlation matrix for the given DataFrame.
    """
    corr = df.corr()
    corr = corr[(abs(corr) > 0.7)]
    plt.figure(figsize=(18, 13))
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', cbar=True)
    plt.title('Correlation Heatmap')
    plt.show()
plot_correlation_matrix(e.drop(columns=['Date', 'HomeTeam', 'AwayTeam', 'FTHG', 'FTAG', 'FTR', 'HTR', 'B365H', 'B365D',
       'B365A', 'PSH', 'PSD', 'PSA',]))
```

```python
# e["Date"] have some formats in DD/MM/YYYY and some formats in DD/MM/YY
# Convert all dates to a consistent format
import re
def convert_date_format(date_str):
    # Check if the date is in DD/MM/YY format
    if re.match(r'^\d{2}/\d{2}/\d{2}$', date_str):
        return pd.to_datetime(date_str, format='%d/%m/%y')
    # Otherwise, assume it's in DD/MM/YYYY format
    return pd.to_datetime(date_str, format='%d/%m/%Y')
e["Date"] = e["Date"].apply(convert_date_format)
# Check for any remaining NaN values in the 'Date' column
if e["Date"].isna().any():
    print("There are still NaN values in the 'Date' column after conversion.")


```
