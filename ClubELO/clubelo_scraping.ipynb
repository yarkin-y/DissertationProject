```python
date_list = understat_matches['date_for_clubelo'].unique().tolist()
```

```python
date_to_df = {}
for date in tqdm.tqdm(date_list, desc="Fetching ClubELO data")  :
    try:
        clubelo_request = requests.get(f"http://api.clubelo.com/{date}")
        if clubelo_request.status_code != 200:
            print(f"Failed to fetch data for {date}: {clubelo_request.status_code}")
            break
        clubelo_df = pd.read_csv(io.StringIO(clubelo_request.text))
        date_to_df[date] = clubelo_df
    except Exception as e:
        print(f"Error fetching data for {date}: {e}")
```

```python
team_date_elo_df = pd.DataFrame()
for date, df in tqdm.tqdm(date_to_df.items()):
    df["date"] = date
    team_date_elo_df = pd.concat([team_date_elo_df, df], ignore_index=True)
```

```python
team_date_elo_df.to_csv("Data/clubelo_team_date_elo.csv", index=False)
```

```python
understat_matches = pd.merge(understat_matches, team_date_elo_df[['Club', 'date', 'Elo']], how='left', left_on=['team_h', 'date_for_clubelo'], right_on=['Club', 'date'])
```

```python
def nan_analysis(df):
    # Returns teams (team_h) with NaN values in Elo rating
    return df[df['Elo'].isna()]['team_h'].unique()
```
