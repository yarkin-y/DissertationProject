```python
# Check if team_h names match home_team names in sofa_indexes_df
# Show which teams are not matching
# Show for both sofa and understat indexes
def check_team_names(s, u):
    s_teams = set(s['team_h'].unique())
    u_teams = set(u['team_h'].unique())
    return s_teams - u_teams
sofa_team_names = sofa_indexes_df[['team_h']].drop_duplicates()
understat_team_names = understat_indexes_df[['team_h']].drop_duplicates()
non_matching_teams_sofa = check_team_names(sofa_team_names, understat_team_names)
non_matching_teams_understat = check_team_names(understat_team_names, sofa_team_names)
if non_matching_teams_sofa:
    print("Non-matching teams found in Sofa:", non_matching_teams_sofa)
if non_matching_teams_understat:
    print("Non-matching teams found in Understat:", non_matching_teams_understat)
```

```python
import re
# Find columns that have _home or _away in their names
def pivot_dataframe(df):
    home_cols = [col for col in df.columns if re.search(r'_home$', col)]
    # Create home DataFrame and rename columns to their original names without _home suffix

    home_df = df[["id", "sofa_id", "h", "date"] + home_cols].copy()
    home_df.columns = [col.replace('_home', '') for col in home_df.columns]
    home_df.rename(columns={'h': 'team_id'}, inplace=True)
    home_df["h_a"]= "h"


    away_cols = [col for col in df.columns if re.search(r'_away$', col)]
    away_df = df[["id", "sofa_id", "a", "date"] + away_cols].copy()
    away_df.columns = [col.replace('_away', '') for col in away_df.columns]
    away_df.rename(columns={'a': 'team_id'}, inplace=True)
    away_df["h_a"]= "a"
    return pd.concat([home_df, away_df], ignore_index=True)
df_to_save_pivoted = pivot_dataframe(df_to_save)
df_to_save_pivoted
```
