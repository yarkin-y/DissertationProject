```python
def pivot_dataframe(df, features_to_pivot):
    home_df= df[df["h_a"]== "h"].copy()
    home_df.rename(columns={"team_id":"h"}, inplace=True)
    home_df.rename(columns={feature: f"{feature}_h" for feature in features_to_pivot}, inplace=True)
    home_df.drop(columns=["h_a"], inplace=True)
    away_df= df[df["h_a"]== "a"].copy()
    away_df.rename(columns={"team_id":"a"}, inplace=True)
    away_df.rename(columns={feature: f"{feature}_a" for feature in features_to_pivot}, inplace=True)
    if "date" in away_df.columns:
        away_df.drop(columns=["h_a","date"], inplace=True)
    else: 
        away_df.drop(columns=["h_a"], inplace=True)
    merged_df = pd.merge(home_df, away_df, on="id")
    for feature in features_to_pivot:
        merged_df[feature] = merged_df[f"{feature}_h"] - merged_df[f"{feature}_a"]
    merged_df.drop(columns=[f"{feature}_h" for feature in features_to_pivot] + [f"{feature}_a" for feature in features_to_pivot], inplace=True)
    return merged_df
```

```python
import re
import pandas as pd

def map_window_features(original_features, window_features):
    """
    Map original feature names to their corresponding window-aggregated variants
    using regex.
    """
    mapping = {}
    for orig in original_features:
        # Build a regex that finds window features starting with orig + underscore
        pattern = re.compile(rf"^{re.escape(orig)}_")
        match = [w for w in window_features if pattern.match(w)]
        
        if match:
            # If multiple matches exist, take the first
            mapping[orig] = match[0]
        else:
            mapping[orig] = None  # No match found
    return mapping


def compute_window_errors(pivoted_df, original_features, window_features):
    """
    Compute absolute errors between original and window-aggregated features.
    """
    mapping = map_window_features(original_features, window_features)
    error_df = pd.DataFrame()

    for orig, win in mapping.items():
        if win and orig in pivoted_df.columns and win in pivoted_df.columns:
            error_df[orig + "_error"] = (pivoted_df[orig] - pivoted_df[win]).abs()

    return error_df.mean().reset_index(name="mean_error").rename(columns={"index": "feature"})
```

```python
def improvement(df, org, win, mod):
    win_df = compute_window_errors(df, org, win)
    mod_df = compute_window_errors(df, org, mod)
    improvement_df = pd.DataFrame()
    improvement_df["feature"] = win_df["feature"]
    improvement_df["percentage_improvement"] = ((win_df["mean_error"] - mod_df["mean_error"]) / win_df["mean_error"]) * 100
    return improvement_df
```
